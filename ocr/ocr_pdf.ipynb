```python
!pip install -q requests pymupdf

import fitz  # PyMuPDF
import requests

URL = 'PASTE_YOUR_FUNCTION_URL_HERE'
PDF = 'file.pdf'  # your PDF (upload it next to this notebook)

pages = []
doc = fitz.open(PDF)
for i, page in enumerate(doc, start=1):
    png = page.get_pixmap(dpi=200).tobytes('png')  # render page to an image
    r = requests.post(URL, data=png,
                      headers={'Content-Type': 'application/octet-stream'}, timeout=90)
    text = r.json()['text'] if r.ok else f'ERROR {r.status_code}: {r.text[:200]}'
    pages.append(f'--- Page {i} ---\n{text}')
    print(f'page {i}/{len(doc)} done')

full_text = '\n\n'.join(pages)
open('output.txt', 'w').write(full_text)
print(full_text[:1000])
```
